```python
import pandas as pd
import plotly.express as px
df = pd.read_csv('../data/hajj_pilgrims_raw.csv')
df.head()
```

```python
fig = px.line(df, x='year', y='pilgrims_total', title='Hajj Pilgrims Over Time (1970-2025)', markers=True)
fig.show()
```

```python
fig.add_vline(x=2020, line_dash='dash', line_color='red')
fig.add_annotation(x=2020, y=3100000, text='COVID-19 drop', showarrow=False, textangle=0)
fig.add_vline(x=2010, line_dash='dash', line_color='gray')
fig.add_annotation(x=2010, y=3700000, text='Haramain railway + Jeddah airport expansion', showarrow=False, textangle=0)
fig.update_layout(margin=dict(t=150))
fig.show()
fig.write_html('../hajj_chart.html', include_plotlyjs='cdn')
```
